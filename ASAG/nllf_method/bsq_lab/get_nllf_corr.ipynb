```python
import pandas as pd
import json

dataset_nllfg = pd.read_excel("dataset_juke.xlsx", index_col=0)

bsq_corr = dataset_nllfg.groupby("b").apply(lambda x: x["label"].corr(x["label_task"])).to_dict()
ci_corr = {}
for ci, v in bsq_corr.items():
    v = v if str(v) != "nan" else 0
    for a in "YN":
        c = f"{ci} ({a})"
        ci_corr[c] = v if a == "Y" else -v

dataset_nllfg.groupby("b").apply(lambda x: x["bsq"].iloc[0]).to_dict(), bsq_corr
```

```python
# Save the dictionary as a JSON file
with open("output/corr.json", "w") as json_file:
    json.dump(ci_corr, json_file, indent=4)
```
