```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Si" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c2 (Y)**: "La respuesta muestra un cálculo correcto de una cantidad."
- **c2 (N)**: "La respuesta no muestra un cálculo correcto de una cantidad."
- **c7 (Y)**: "La respuesta explica por qué un personaje está equivocado."
- **c7 (N)**: "La respuesta no explica por qué un personaje está equivocado."
- **c4 (Y)**: "La respuesta está libre de errores conceptuales."
- **c4 (N)**: "La respuesta contiene errores conceptuales."
- **c5 (Y)**: "La respuesta muestra una comprensión correcta de la pregunta."
- **c5 (N)**: "La respuesta no muestra una comprensión correcta de la pregunta."
- **c0 (Y)**: "La respuesta indica correctamente una cantidad."
- **c0 (N)**: "La respuesta no indica correctamente una cantidad."
- **c10 (Y)**: "La respuesta está escrita de manera que se pueda entender fácilmente."
- **c10 (N)**: "La respuesta no está escrita de manera que se pueda entender fácilmente."
- **c6 (Y)**: "La respuesta está redactada de forma clara y concisa."
- **c6 (N)**: "La respuesta no está redactada de forma clara y concisa."
- **c1 (Y)**: "La respuesta está correctamente escrita en formato numérico."
- **c1 (N)**: "La respuesta no está correctamente escrita en formato numérico."
- **c9 (Y)**: "La respuesta está acompañada de una explicación."
- **c9 (N)**: "La respuesta no está acompañada de una explicación."
- **c12 (Y)**: "La respuesta está completa y no falta ninguna información relevante."
- **c12 (N)**: "La respuesta está incompleta o falta información relevante."
- **c8 (Y)**: "La respuesta aborda la pregunta."
- **c8 (N)**: "La respuesta no aborda la pregunta."
- **c3 (Y)**: "La respuesta identifica correctamente el valor."
- **c3 (N)**: "La respuesta no identifica correctamente el valor."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
