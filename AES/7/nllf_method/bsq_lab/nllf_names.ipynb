```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c8 (Y)**: "The essay provides an example of patience."
- **c8 (N)**: "The essay does not provide an example of patience."
- **c11 (Y)**: "The essay describes the experience of the author or someone else."
- **c11 (N)**: "The essay does not describe the experience of the author or someone else."
- **c7 (Y)**: "The essay describes a situation where patience was required."
- **c7 (N)**: "The essay does not describe a situation where patience was required."
- **c9 (Y)**: "The essay demonstrates understanding and tolerance in the face of difficulties."
- **c9 (N)**: "The essay does not demonstrate understanding and tolerance in the face of difficulties."
- **c3 (Y)**: "The essay shows that the person in the story did not complain during the difficult situation."
- **c3 (N)**: "The essay does not show that the person in the story did not complain during the difficult situation."
- **c10 (Y)**: "The essay has a clear focus on the topic throughout."
- **c10 (N)**: "The essay does not have a clear focus on the topic throughout."
- **c1 (Y)**: "The essay has a clear and focused central idea."
- **c1 (N)**: "The essay does not have a clear and focused central idea."
- **c6 (Y)**: "The essay has clear and logical connections between ideas and/or events."
- **c6 (N)**: "The essay does not have clear and logical connections between ideas and/or events."
- **c5 (Y)**: "The essay has a logical order to the ideas and/or events presented."
- **c5 (N)**: "The essay does not have a logical order to the ideas and/or events presented."
- **c4 (Y)**: "The essay uses a variety of sentence structures effectively."
- **c4 (N)**: "The essay does not use a variety of sentence structures effectively."
- **c0 (Y)**: "The essay uses clear and effective language throughout."
- **c0 (N)**: "The essay does not use clear and effective language throughout."
- **c2 (Y)**: "The essay consistently uses correct grammar, usage, spelling, capitalization, and punctuation for the grade level."
- **c2 (N)**: "The essay does not consistently use correct grammar, usage, spelling, capitalization, and punctuation for the grade level."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
