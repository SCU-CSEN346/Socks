```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c7 (Y)**: "The essay clearly states the author's opinion on the effects of computers on people."
- **c7 (N)**: "The essay does not clearly state the author's opinion on the effects of computers on people."
- **c4 (Y)**: "The essay shows critical thinking and analysis."
- **c4 (N)**: "The essay does not show critical thinking and analysis."
- **c5 (Y)**: "The essay elaborates on the reasons given."
- **c5 (N)**: "The essay does not elaborate on the reasons given."
- **c0 (Y)**: "The essay has a clear introduction."
- **c0 (N)**: "The essay does not have a clear introduction."
- **c11 (Y)**: "The essay demonstrates clarity in its ideas."
- **c11 (N)**: "The essay does not demonstrate clarity in its ideas."
- **c9 (Y)**: "The essay flows logically from one idea to the next."
- **c9 (N)**: "The essay does not flow logically from one idea to the next."
- **c2 (Y)**: "The essay uses language and tone appropriate for the audience."
- **c2 (N)**: "The essay does not use language and tone appropriate for the audience."
- **c8 (Y)**: "The essay has grammatical or spelling errors that interfere with its clarity."
- **c8 (N)**: "The essay does not have grammatical or spelling errors that interfere with its clarity."
- **c1 (Y)**: "The essay demonstrates an understanding of the intended audience."
- **c1 (N)**: "The essay does not demonstrate an understanding of the intended audience."
- **c6 (Y)**: "The essay explains how the supporting evidence supports its position."
- **c6 (N)**: "The essay does not explain how the supporting evidence supports its position."
- **c10 (Y)**: "The essay avoids using jargon or technical terms that may confuse the audience."
- **c10 (N)**: "The essay does not avoid using jargon or technical terms that may confuse the audience."
- **c3 (Y)**: "The essay provides support for its position."
- **c3 (N)**: "The essay does not provide support for its position."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
