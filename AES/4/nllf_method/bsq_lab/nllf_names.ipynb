```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c7 (Y)**: "The essay avoids restating the conclusion without providing additional insights or analysis."
- **c7 (N)**: "The essay does not avoid restating the conclusion without providing additional insights or analysis."
- **c2 (Y)**: "The essay uses information explicitly stated in the text."
- **c2 (N)**: "The essay does not use information explicitly stated in the text."
- **c11 (Y)**: "The essay uses this information accurately and relevantly."
- **c11 (N)**: "The essay does not use this information accurately and relevantly."
- **c6 (Y)**: "The essay clarifies concepts beyond the literal meaning."
- **c6 (N)**: "The essay does not clarify concepts beyond the literal meaning."
- **c5 (Y)**: "The essay connects ideas in a meaningful way."
- **c5 (N)**: "The essay does not connect ideas in a meaningful way."
- **c9 (Y)**: "The essay goes beyond the surface level of the subject matter."
- **c9 (N)**: "The essay does not go beyond the surface level of the subject matter."
- **c1 (Y)**: "The essay demonstrates a clear understanding of the text."
- **c1 (N)**: "The essay does not demonstrate a clear understanding of the text."
- **c3 (Y)**: "The essay shows evidence of making connections between the text and the world beyond."
- **c3 (N)**: "The essay does not show evidence of making connections between the text and the world beyond."
- **c0 (Y)**: "The essay connects the support to the conclusion."
- **c0 (N)**: "The essay does not connect the support to the conclusion."
- **c10 (Y)**: "The essay offers an explanation that clarifies the relationship between the text and the question."
- **c10 (N)**: "The essay does not offer an explanation that clarifies the relationship between the text and the question."
- **c4 (Y)**: "The essay omits important details that are necessary to understand the text."
- **c4 (N)**: "The essay does not omit important details that are necessary to understand the text."
- **c8 (Y)**: "The essay fails to provide a coherent explanation of the text in relation to the question."
- **c8 (N)**: "The essay does not fail to provide a coherent explanation of the text in relation to the question."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
