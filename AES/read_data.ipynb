```python
import pandas as pd
from pathlib import Path
```

```python
RANDOM_SEED = 42
```

```python
raw_dataset = pd.read_csv("data/train.tsv", sep='\t', encoding='latin1')
raw_dataset
```

```python
pd.set_option('display.max_columns', None)
```

```python
ix = 1

essay_set = raw_dataset[raw_dataset["essay_set"] == ix]
essay_set = essay_set.dropna(axis=1)
essay_set = essay_set.drop(columns="rater1_domain1	rater2_domain1".split())
essay_set = essay_set.rename(columns={"domain1_score": "score"})

essay_set = essay_set.sample(frac=1, random_state=RANDOM_SEED).reset_index(drop=True)
split_index = int(0.8 * len(essay_set))
essay_set['split'] = ['train'] * split_index + ['test'] * (len(essay_set) - split_index)
essay_set = essay_set.sample(frac=1, random_state=RANDOM_SEED).reset_index(drop=True)

essay_set.to_csv("data/essay_set_1.csv")
```
