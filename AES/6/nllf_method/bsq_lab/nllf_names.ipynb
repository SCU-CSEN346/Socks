```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c0 (Y)**: "The essay clearly describes the obstacles builders faced in docking dirigibles."
- **c0 (N)**: "The essay does not clearly describe the obstacles builders faced in docking dirigibles."
- **c4 (Y)**: "The essay includes relevant information from the excerpt."
- **c4 (N)**: "The essay does not include relevant information from the excerpt."
- **c9 (Y)**: "The essay completely covers the challenges encountered in building the dock."
- **c9 (N)**: "The essay does not completely cover the challenges encountered in building the dock."
- **c7 (Y)**: "The essay avoids relying solely on general summaries without elaboration."
- **c7 (N)**: "The essay relies solely on general summaries without elaboration."
- **c6 (Y)**: "The essay provides information about the impact of the obstacles on the final design of the building."
- **c6 (N)**: "The essay does not provide information about the impact of the obstacles on the final design of the building."
- **c5 (Y)**: "The essay demonstrates a clear understanding of the excerpt."
- **c5 (N)**: "The essay does not demonstrate a clear understanding of the excerpt."
- **c3 (Y)**: "The essay shows a thorough engagement with the excerpt."
- **c3 (N)**: "The essay does not show a thorough engagement with the excerpt."
- **c2 (Y)**: "The essay avoids adding unnecessary or irrelevant information."
- **c2 (N)**: "The essay includes unnecessary or irrelevant information."
- **c10 (Y)**: "The essay addresses the prompt or topic."
- **c10 (N)**: "The essay does not address the prompt or topic."
- **c1 (Y)**: "The essay provides relevant examples or evidence."
- **c1 (N)**: "The essay does not provide relevant examples or evidence."
- **c11 (Y)**: "The essay maintains a clear focus throughout."
- **c11 (N)**: "The essay does not maintain a clear focus throughout."
- **c8 (Y)**: "The essay demonstrates a consistent and coherent argument."
- **c8 (N)**: "The essay does not demonstrate a consistent and coherent argument."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
