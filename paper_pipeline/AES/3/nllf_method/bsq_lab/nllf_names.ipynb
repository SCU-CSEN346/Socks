```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c8 (Y)**: "The essay identifies features of the setting that affect the cyclist."
- **c8 (N)**: "The essay does not identify features of the setting that affect the cyclist."
- **c1 (Y)**: "The essay addresses all parts of the question."
- **c1 (N)**: "The essay does not address all parts of the question."
- **c5 (Y)**: "The essay avoids irrelevant information or arguments."
- **c5 (N)**: "The essay includes irrelevant information or arguments."
- **c6 (Y)**: "The essay avoids using misleading or incorrect information from the text."
- **c6 (N)**: "The essay includes misleading or incorrect information from the text."
- **c4 (Y)**: "The essay clarifies concepts beyond the literal."
- **c4 (N)**: "The essay does not clarify concepts beyond the literal."
- **c10 (Y)**: "The essay provides new insights."
- **c10 (N)**: "The essay does not provide new insights."
- **c9 (Y)**: "The essay goes beyond the surface level of the topic."
- **c9 (N)**: "The essay does not go beyond the surface level of the topic."
- **c2 (Y)**: "The essay demonstrates a clear understanding of the text."
- **c2 (N)**: "The essay does not demonstrate a clear understanding of the text."
- **c3 (Y)**: "The essay connects the support to the conclusion."
- **c3 (N)**: "The essay does not connect the support to the conclusion."
- **c0 (Y)**: "The essay demonstrates an ability to make connections between the text and its context."
- **c0 (N)**: "The essay does not demonstrate an ability to make connections between the text and its context."
- **c11 (Y)**: "The essay avoids relying on superficial or inaccurate information about the text."
- **c11 (N)**: "The essay relies on superficial or inaccurate information about the text."
- **c7 (Y)**: "The essay shows evidence of critical thinking and analysis."
- **c7 (N)**: "The essay does not show evidence of critical thinking and analysis."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
