```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c0 (Y)**: "The essay describes a true story."
- **c0 (N)**: "The essay does not describe a true story."
- **c7 (Y)**: "The essay includes laughter as an element or part of the story."
- **c7 (N)**: "The essay does not include laughter as an element or part of the story."
- **c11 (Y)**: "The essay avoids using cliches, repetition, or irrelevant information."
- **c11 (N)**: "The essay uses cliches, repetition, or irrelevant information."
- **c6 (Y)**: "The essay clearly and consistently maintains focus on a single topic throughout."
- **c6 (N)**: "The essay does not maintain focus on a single topic throughout."
- **c2 (Y)**: "The essay clearly and effectively communicates the main ideas."
- **c2 (N)**: "The essay does not effectively communicate the main ideas."
- **c8 (Y)**: "The essay's voice matches its topic, purpose, and audience."
- **c8 (N)**: "The essay's voice does not match its topic, purpose, and audience."
- **c4 (Y)**: "The essay shows a clear commitment to the topic."
- **c4 (N)**: "The essay does not show a clear commitment to the topic."
- **c1 (Y)**: "The essay uses a varied and well-chosen vocabulary to effectively convey its message."
- **c1 (N)**: "The essay does not use a varied and well-chosen vocabulary."
- **c5 (Y)**: "The essay has a limited vocabulary or frequent misuse of words that obscure the meaning."
- **c5 (N)**: "The essay does not have a limited vocabulary or misuse of words."
- **c10 (Y)**: "The essay has occasional awkward constructions that may require the reader to slow down or reread."
- **c10 (N)**: "The essay does not have awkward constructions that impede readability."
- **c3 (Y)**: "The essay has a clear and coherent structure."
- **c3 (N)**: "The essay does not have a clear and coherent structure."
- **c9 (Y)**: "The essay has frequent and significant errors in standard writing conventions that impede readability."
- **c9 (N)**: "The essay does not have frequent and significant errors in standard writing conventions."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
