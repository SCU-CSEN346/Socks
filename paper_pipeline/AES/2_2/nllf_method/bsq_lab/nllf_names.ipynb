```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c4 (Y)**: "The essay takes a clear position on whether or not to remove offensive materials from libraries."
- **c4 (N)**: "The essay does not take a clear position on whether or not to remove offensive materials from libraries."
- **c10 (Y)**: "The essay uses logical reasoning to support its position on censorship."
- **c10 (N)**: "The essay does not use logical reasoning to support its position on censorship."
- **c5 (Y)**: "The essay fully accomplishes the given task."
- **c5 (N)**: "The essay does not fully accomplish the given task."
- **c9 (Y)**: "The essay provides a thorough analysis or explanation."
- **c9 (N)**: "The essay does not provide a thorough analysis or explanation."
- **c6 (Y)**: "The essay has a distinctive quality that sets it apart as outstanding."
- **c6 (N)**: "The essay does not have a distinctive quality that sets it apart as outstanding."
- **c1 (Y)**: "The essay has a clear and logical organization."
- **c1 (N)**: "The essay does not have a clear and logical organization."
- **c2 (Y)**: "The essay uses appropriate language and tone for the intended audience and purpose."
- **c2 (N)**: "The essay does not use appropriate language and tone for the intended audience and purpose."
- **c0 (Y)**: "The essay includes a sufficient number of relevant ideas."
- **c0 (N)**: "The essay does not include a sufficient number of relevant ideas."
- **c8 (Y)**: "The essay demonstrates basic writing techniques such as proper grammar, punctuation, and sentence structure."
- **c8 (N)**: "The essay does not demonstrate basic writing techniques such as proper grammar, punctuation, and sentence structure."
- **c7 (Y)**: "The essay maintains a consistent length of responses without being overly brief or rambling and repetitive."
- **c7 (N)**: "The essay does not maintain a consistent length of responses or is overly brief or rambling and repetitive."
- **c3 (Y)**: "The essay flows smoothly without any abrupt changes in the direction of the argument."
- **c3 (N)**: "The essay does not flow smoothly and has abrupt changes in the direction of the argument."
- **c11 (Y)**: "The essay fails to support its opinion, summarize effectively, tell a coherent story, or present clear information."
- **c11 (N)**: "The essay does not fail in supporting its opinion, summarizing effectively, telling a coherent story, or presenting clear information."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
