```python
import pandas as pd
import json
```

```python
## BSQS
bsqs = pd.read_csv("output/clustered_bsqs.csv", index_col=0)
bsqs["id"] = bsqs["cluster"].apply(lambda x: f"c{x}")
bsqs = bsqs[bsqs["is_centroid"]].reset_index().set_index("id")["bsq"].to_dict()
bsqs
```

```python
named_nllf = {
    f"{k} ({a})": f"{v} Yes" if a == "Y" else f"{v} No"
    for k, v in bsqs.items()
    for a in "YN"
}
named_nllf
```

```python
transformed_names = """- **c5 (Y)**: "The essay clearly describes the mood created by the author."
- **c5 (N)**: "The essay does not clearly describe the mood created by the author."
- **c4 (Y)**: "The essay avoids using irrelevant or generic information."
- **c4 (N)**: "The essay includes irrelevant or generic information."
- **c0 (Y)**: "The essay demonstrates a clear understanding of the memoir."
- **c0 (N)**: "The essay does not demonstrate a clear understanding of the memoir."
- **c8 (Y)**: "The essay contains errors or inconsistencies in describing the mood."
- **c8 (N)**: "The essay does not contain errors or inconsistencies in describing the mood."
- **c7 (Y)**: "The essay fails to make connections between the memoir and the topic."
- **c7 (N)**: "The essay does not fail to make connections between the memoir and the topic."
- **c6 (Y)**: "The essay explains how the examples reflect the mood."
- **c6 (N)**: "The essay does not explain how the examples reflect the mood."
- **c3 (Y)**: "The essay uses descriptive language to create a mood."
- **c3 (N)**: "The essay does not use descriptive language to create a mood."
- **c9 (Y)**: "The essay maintains a consistent mood throughout."
- **c9 (N)**: "The essay does not maintain a consistent mood throughout."
- **c1 (Y)**: "The essay includes information from the memoir."
- **c1 (N)**: "The essay does not include information from the memoir."
- **c10 (Y)**: "The essay lacks misinterpretations."
- **c10 (N)**: "The essay contains misinterpretations."
- **c2 (Y)**: "The essay provides relevant examples or evidence."
- **c2 (N)**: "The essay does not provide relevant examples or evidence."
- **c11 (Y)**: "The essay shows a minimal effort to complete the task."
- **c11 (N)**: "The essay does not show a minimal effort to complete the task."""

transformed_names = {row[row.index("- **")+4:row.index("**:")].strip(): row[row.index("**: ")+5:].replace('"', "").replace(".", "") for row in transformed_names.split("\n")}
transformed_names
```

```python
# Save the dictionary as a JSON file
with open("output/transformed_names.json", "w") as json_file:
    json.dump(transformed_names, json_file, indent=4)
```
